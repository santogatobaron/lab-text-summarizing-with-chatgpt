## Setup

````python
from openai import OpenAI
import os

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')

# check get key
if OPENAI_API_KEY:
    print("API key loaded successfully")
else:
    print("API key NOT found")
````

````python
client = OpenAI()


def get_completion(prompt, model="gpt-3.5-turbo"): # Andrew mentioned that the prompt/ completion paradigm is preferable for this class
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0, # this is the degree of randomness of the model's output
    )
    return response.choices[0].message.content
````

## Summarize with a word/sentence/character limit

````python
prompt = f"""
Your task is to generate a short summary of a product \
review from an ecommerce site. 

Summarize the review below, delimited by triple 
backticks, in at most 30 words. 

Review: ```{prod_review}```
"""

response = get_completion(prompt)
print(response)
````

1. Summarized and quantified ratings

````python
def summarize_reviews(reviews_list):
    
    # Construimos el bloque de texto con todas las reviews
    reviews_text = ""
    for i, review in enumerate(reviews_list, start=1):
        reviews_text += f"Review {i}: {review}\n\n"
    
    system_prompt = """
You are an expert product review analyst.

Your task:
- Read all the reviews.
- For EACH review, provide:
    - A short summary (maximum 7 words ).
    - Sentiment: positive, negative, or neutral.
    - Rating: from 0 to 10 based on the opinion.

Formatting rules:
- Do NOT merge reviews.
- Follow exactly this structure:

Review i: summary (max 7 words)
Sentiment: positive, negative or neutral
Rating: X/10

"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": reviews_text}
    ]
    
    response = get_completion_from_messages(messages)
    
    return print(response)

summarize_reviews(reviews)
````

````python
def summarize_overall_opinion(reviews_list):
    
    reviews_text = ""
    for i, review in enumerate(reviews_list, start=1):
        reviews_text += f"Review {i}: {review}\n\n"
    
    system_prompt = """
You are an expert product review analyst.

Your task:
- Read all the reviews.
- Write exactly around 30 words summarizing the overall customer opinion.
- Mention general agreement or recurring themes.
- Do NOT summarize each review separately.
- Produce one single paragraph.
"""
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": reviews_text}
    ]
    
    response = get_completion_from_messages(messages, temperature=0)
    
    return print(response)

summarize_overall_opinion(reviews)
````

3. Extract opinions that need to be dealt with

````python
def analyze_extreme_reviews(reviews_list):
    
    reviews_text = ""
    for i, review in enumerate(reviews_list, start=1):
        reviews_text += f"Review {i}: {review}\n\n"
    
    system_prompt = """
You are an expert product review analyst.

Your task:
1. Read all reviews.
2. Estimate a rating from 0 to 10 for each review.
3. Select ONLY reviews that you consider:
   - Perfect (10/10)
   - Or below 5/10
4. For each selected review, provide:

- Rating: X/10
- Summary: approximately 10 words summarizing the opinion
- Main problem: (only if negative, below 5/10)
- Response message: approximately 10 words
    * If positive: short thank-you message
    * If negative: short message thanking for feedback and acknowledging issue

Formatting rules:
- Do NOT include reviews that are between 5 and 9.
- Keep responses concise.
- Follow the structure exactly.
- Separate each review with a blank line.
"""
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": reviews_text}
    ]
    
    response = get_completion_from_messages(messages, temperature=0)
    
    return print(response)

analyze_extreme_reviews(reviews)

````
